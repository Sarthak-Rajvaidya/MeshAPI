```python
import os
from getpass import getpass
from dotenv import load_dotenv
load_dotenv()
```

```python
from meshapi import MeshAPI
MESHAPI_TOKEN = os.getenv("MESH_API_KEY") or getpass("MeshAPI token (rsk_...): ")
client = MeshAPI(base_url=os.getenv("MESHAPI_BASE_URL", "https://api.meshapi.ai"), token=MESHAPI_TOKEN)
print("Client ready.")
```

```python
len(client.models.list())
```

```python
ALL_MODELS = {m.id: m for m in client.models.list()}
print(f"{len(ALL_MODELS)} models available through this key")
```

```python
#HIT THE MESHAPI CLIENT
from meshapi import MeshAPI, ChatCompletionParams,ChatMessage

client = MeshAPI(
    base_url="https://api.meshapi.ai",
    token=os.getenv("MESH_API_KEY")
)



resp = client.chat.completions.create(
    ChatCompletionParams(
        model="auto",
        messages=[
            ChatMessage(
                role="user",
                content="In one sentence, what is an AI gateway?"
            )
        ],
        max_tokens=60
    )
)

print(resp.choices[0].message.content)
```
